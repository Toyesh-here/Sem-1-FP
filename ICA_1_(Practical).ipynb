#SET-2, 10 marks, Q1
Write a python program that accepts a person's weight(kg) and height(m), calculate their BMI using the formula BMI=weight/height**2, and displays the BMI(rounded to 2 decimals)  along with the category: underweight(<18.5), normal(18.5-24.9), overweight(24.9-29.9), or obese(>=30.0).


```python
weight=float(input("Enter person's weight (in kg): "))
height=float(input("Enter person's height (in m): "))

if weight<=0 or height<=0 :
  print("ERROR! : Weight and height must be positive values")
else :
  bmi=(weight)/(height**2)
  bmi=round(bmi,2)
  if bmi < 18.5 :
    category="Underweight"
  elif bmi <= 24.9 :
    category="Normal"
  elif bmi<= 29.9 :
    category="Overweight"
  else :
    category="Obese"
  print("BMI: ", bmi)
  print("Category: ", category)
```

#Set-2, 10 marks, Q2
Two sets, PythonClub and ExcelClub, contain the names of students enrolled in each club. Write a python program to display: (a) Student in both te clubs, (b) students enrolled only in PythonClub, and (c) whether the two sets are disjoint.

```python
PythonClub={"Aarav","Krishna","Keshav","Riya","Navya"}
ExcelClub={"Riya","Krishna","Aastha","Vyom"}

Both=PythonClub & ExcelClub
Only_PythonClub=PythonClub - ExcelClub
Disjoint=PythonClub.isdisjoint(ExcelClub)

print("Students in both clubs: ", Both)
print("Students only in python club: ", Only_PythonClub)
print("Are the two sets disjoint ? -->", Disjoint)
```
