# ***Q1***
[Python Basics & Data Structures] Using the sep and end parameters of print(), display the words Mon Tue Wed
separated by " | ", and make a second print("Thu") continue on the SAME line after the text " -> ". Add a comment
explaining what the end parameter does.



```python
# end decides what print() writes after its last item.
# By default end="\n" (a new line), so the next print starts on a new line.
# Setting end=" -> " replaces the newline, so the next print continues on the same line.
print("Mon", "Tue", "Wed", sep=" | ", end=" -> ")
print("Thu")
```

# ***Q2***
[Strings, I/O & Operators] Let price = 1234.5678. Using an f-string, print it rounded to 2 decimals with a thousands
separator and a "Rs" prefix, i.e., Rs 1,234.57. [5 Marks]

```python
price = 1234.5678
print(f"Rs {price:,.2f}")
```

# ***Q3***
[Control Flow & Loops] Using a while loop and break, find and print the first number whose square exceeds 200.
[5 Marks]

```python
n = 1
while True:
    if n * n > 200:
        print("First number whose square exceeds 200:", n)
        break
    n += 1
```

# ***Q4***
[Functions] Write two functions: area_return(l, b) that RETURNS the area of a rectangle and area_print(l, b) that
only PRINTS it. Print the result of calling each, and explain in a comment why one of them shows None. [5 Marks]


```python
def area_return(l, b):
    return l * b

def area_print(l, b):
    print(l * b)

print(area_return(5, 4))
print(area_print(5, 4))

# area_print() has no return statement, so Python automatically returns None.
# Calling it prints 20 from inside the function, then the outer print() shows
# the returned value, which is None.
```
