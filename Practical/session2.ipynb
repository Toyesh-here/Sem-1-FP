```python
no_of_elements= int(input("Enter the number of elements: "))
list_of_elements = []
for i in range(no_of_elements):
    number = int(input("Enter element : "))
    list_of_elements.append(number)
print("List of elements:", list_of_elements)
list_of_elements.sort()
print("the min and max is: ", list_of_elements[0], "and", list_of_elements[-1])
```
