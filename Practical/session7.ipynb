SAME CODE IN WHILE LOOP

```python
L1=[['Python'],['Program'],['Example']]

i = 0
while i < len(L1):
    j = 0
    while j < len(L1[i]):
        k = 0
        while k < len(L1[i][j]):
            print(L1[i][j][k])
            k += 1
        j += 1
    print()
    i += 1
```

```python
num = int(input("Enter a number: "))
sum = 0
while num > 0:
    digit = num % 10
    sum += digit
    num //= 10
print(f"The sum of digits is: {sum}")
```

```python
num = int(input("Enter an integer: "))

if num >= 0:
    print(f"The number {num} is positive.")
    if num % 2 == 0:
        print(f"The number {num} is even.")
    else:
        print(f"The number {num} is odd.")
else:
    print(f"The number {num} is negative.")
```
